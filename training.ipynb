```python
import os
from ultralytics import YOLO

if not os.path.exists(root_dir):
    print(f"Error: The directory '{root_dir}' does not exist. Please check your Google Drive path.")

    if os.path.exists('/content/gdrive/My Drive'):
        print("Contents of 'My Drive':", os.listdir('/content/gdrive/My Drive'))
else:
    yaml_path = os.path.join(root_dir, "data.yaml")
    if not os.path.exists(yaml_path):
        print(f"Error: 'data.yaml' not found in '{root_dir}'. Available files:")
        print(os.listdir(root_dir))
    else:
        print("data.yaml found! Starting training...")
        model = YOLO("yolov8n.yaml")
        results = model.train(data=yaml_path, epochs=80)
```
